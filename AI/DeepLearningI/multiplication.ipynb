#入力を倍にして出力
最初は単純パーセプトロンを題材に簡単なニューラルネットを実装します。<BR>
課題は入力された値を倍にする事を学習によって獲得します。<BR>
学習データは、入力値と入力値を倍にした値を正解データとして与えます。<br>

```python
# Kerasのインポート
#モデル構築の方法としてSequentialを使う
from tensorflow.keras.models import Sequential
#全結合ネットワーク、活性化関数をインポート
from tensorflow.keras.layers import Dense, Activation
#最適化関数をインポート
from tensorflow.keras.optimizers import SGD
#numpy配列を利用するためインポート
import numpy as np
```

```python
#学習データ X=入力値、Y=正解ラベル
X = np.array([1,2,3,4,5,6,3,1,5,5.5])
Y = np.array([2,4,6,8,10,12,6,2,10,11])
```

```python
#ニューラルネットの定義(Sequentialで行う)
#１入力、1出力の１層を構築
model = Sequential()

#１ノードからなるパーセプトロン
#unitsは出力数で、ノードが1個のため1
#input_shapeは入力となり１個のため1
#Denseは出力ノードと全結合する指定
model.add( Dense( units =  1, input_shape = ( 1, ) ) )

#定義したネットワークをコンパイルしインスタンスを得る
#loss(損失関数)にmean_squared_error(平均二乗誤差)を指定
#optimizer(最適化関数)にSGD(Stochastic Gradient Descent : 確率的勾配降下法)指定
model.compile( loss = 'mean_squared_error', optimizer = 'sgd' )
model.summary()
```

```python
#学習回数30回として学習の実施
#損失関数の値が小さくなっているのを確認
history=model.fit(X, Y, epochs=30)
```

```python
#テストデータを準備
xtest = np.array([7,9,4.2,10, 2.5])
#学習済みモデルに与えて予測させる
result = model.predict(xtest)
```

```python
#結果の確認
for x, r in zip(xtest, result):
    print("入力={}, 出力={:.2f}".format(x, r[0]))
```

```python
#学習によって得られた重み、バイアスを表示し手計算してみましょう。
#レイヤーを指定、今回はひとつしかないため０番目を指定
l = model.layers[0]
#重み、バイアスを取得
params = l.get_weights()
print("重み=", params[0])
print("バイアス=", params[1])
```
