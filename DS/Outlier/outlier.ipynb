```python
# --- データの準備 ---
# 演習データを格納するdatasetsフォルダを作成します
!mkdir -p datasets
# GitHubから演習データ(ice3.csv)をダウンロードし、datasetsフォルダに保存します
!wget -O ./datasets/ice3.csv https://raw.githubusercontent.com/inogit/DXReskill/main/DS/MissingValues/datasets/ice3.csv
```

```python
data = pd.read_csv("./datasets/ice3.csv") # データファイルを読み込み
```

```python
# 読み込んだdataを表示し、中身を確認する
data
```

```python
# 基本統計量の確認（件数、平均、標準偏差、最小/最大、四分位数など）
data.describe()
```
