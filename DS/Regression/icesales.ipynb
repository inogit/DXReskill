```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
!wget https://raw.githubusercontent.com/inogit/DXReskill/refs/heads/main/DS/Regression/datasets/ice2.csv
```

```python
df = pd.read_csv("./dataset/ice2.csv")
```

```python
df.head()
```

```python
x = df.temp.to_frame()
y = df.sales
lr = LinearRegression()
lr.fit(x, y)
```

```python
lr.predict([[30]])
```

```python
plt.scatter(df.temp, df.sales)
plt.plot(df.temp, lr.predict(x))
plt.show()
```

```python
print("y = %.3f * x + %.3f" % (lr.coef_, lr.intercept_))
```
