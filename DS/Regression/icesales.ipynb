```python
# 必要なライブラリをインポートします
import pandas as pd  # データ分析をサポートするライブラリ
from sklearn.linear_model import LinearRegression  # 線形回帰モデルのライブラリ
import matplotlib.pyplot as plt  # グラフ描画ライブラリ
```

```python
# --- データの準備 ---
# 演習データを格納するdatasetsフォルダを作成します
!mkdir -p datasets
# GitHubから演習データ(ice2.csv)をダウンロードし、datasetsフォルダに保存します
!wget -O ./datasets/ice2.csv https://raw.githubusercontent.com/inogit/DXReskill/main/DS/Regression/datasets/ice2.csv
```

```python
# ダウンロードしたCSVファイルをpandasのDataFrameとして読み込みます
df = pd.read_csv("./datasets/ice2.csv")

# 読み込んだデータの先頭5行を表示して、中身を確認します
df.head()
```

```python
# --- モデルの学習 ---
# 説明変数(x)として気温(temp)を、目的変数(y)として売上(sales)を設定します
# scikit-learnのモデルで扱うために、xはDataFrameの形式にしておきます
x = df.temp.to_frame()
y = df.sales
# 線形回帰モデルのインスタンスを作成します
lr = LinearRegression()
# モデルにデータ(x, y)を渡して学習させます
lr.fit(x, y)
```

```python
# 学習済みモデルを使って、気温が30度のときの売上を予測します
lr.predict([[30]])
```

```python
# 実際のデータの散布図を描画します (横軸: 気温, 縦軸: 売上)
plt.scatter(df.temp, df.sales)
# 学習したモデルが予測する直線（回帰直線）を重ねて描画します
plt.plot(df.temp, lr.predict(x))
# グラフを表示します
plt.show()
```

```python
# 学習したモデルが算出した回帰式 y = ax + b の、a (係数/傾き) と b (切片) を表示します
# %.3f は小数点以下3桁まで表示するという意味です
print("y = %.3f * x + %.3f" % (lr.coef_, lr.intercept_))
```
